## Basic data extraction

```python
import pickle

def load_pickle(path: str):
    """
    Load and return data from a pickle file.

    Parameters
    ----------
    path : str
        Path to the .pkl file

    Returns
    -------
    Any
        Python object stored in the pickle file
    """
    with open(path, "rb") as f:
        data = pickle.load(f)

    return data

path = "/workspaces/met4a_data/2026_02_12/1770931655.952085.met4a.pkl"
data = load_pickle(path)
```

## Convert pickle dictionary into pandas df

```python
import pandas as pd

def dict_to_df(data: dict) -> pd.DataFrame:
    """
    Convert a dictionary into a pandas DataFrame.

    Supports:
        - dict of lists  -> columns
        - dict of dicts  -> rows
        - list of dicts  -> rows

    Parameters
    ----------
    data : dict
        Dictionary loaded from pickle

    Returns
    -------
    pd.DataFrame
    """
    if isinstance(data, pd.DataFrame):
        return data  # already a dataframe

    # Case 1: dict of lists (most common)
    try:
        return pd.DataFrame(data)
    except Exception:
        pass

    # Case 2: dict of dicts
    try:
        return pd.DataFrame.from_dict(data, orient="index")
    except Exception:
        pass

    raise ValueError("Unsupported dictionary format for DataFrame conversion.")

path = "/workspaces/met4a_data/2026_02_12/1770931655.952085.met4a.pkl"
data = load_pickle(path)
pickle_df = dict_to_df(data)
```

## Bundle extract from entire directory

```python
from pathlib import Path
import pickle

def load_all_pickles(directory: str | Path) -> list:
    """
    Load all pickle files from a directory into a list.

    Parameters
    ----------
    directory : str | Path
        Path to folder containing .pkl files

    Returns
    -------
    list
        List of objects loaded from pickle files
    """
    directory = Path(directory)

    if not directory.exists():
        raise FileNotFoundError(f"Directory not found: {directory}")

    objects = []

    # Sort so results are deterministic (very helpful for notebooks)
    for pkl_path in sorted(directory.glob("*.pkl")):
        with open(pkl_path, "rb") as f:
            obj = pickle.load(f)
            objects.append(obj)

    return objects

directory_path = "/workspaces/met4a_data/2026_02_12"
data_list = load_all_pickles(directory_path)
```
